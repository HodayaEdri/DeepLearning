```python
STUDENT_ID = 328179361
import torch
torch.manual_seed(STUDENT_ID)
```

```python
from torch import nn
from torch.utils.data import DataLoader
from torchvision import datasets
from torchvision.transforms import ToTensor
```

```python
training_data = datasets.MNIST(
    root="data",
    train=True,
    download=True,
    transform=ToTensor()
)

test_data = datasets.MNIST(
    root="data",
    train=False,
    download=True,
    transform=ToTensor()
)
```

```python
print("Training samples:", len(training_data))
print("Test samples:", len(test_data))

image, label = training_data[0]

print("Image shape:", image.shape)
print("Label:", label)
```

```python
batch_size = 64

train_dataloader = DataLoader(
    training_data,
    batch_size=batch_size,
    shuffle=True
)

test_dataloader = DataLoader(
    test_data,
    batch_size=batch_size,
    shuffle=False
)
```

```python
for X, y in train_dataloader:
    print("Shape of X:", X.shape)
    print("Shape of y:", y.shape)
    print("Data type of y:", y.dtype)
    break
```

```python
class NeuralNetwork(nn.Module):

    def __init__(self):
        super().__init__()

        self.flatten = nn.Flatten()

        self.linear_relu_stack = nn.Sequential(
            nn.Linear(28 * 28, 512),
            nn.ReLU(),

            nn.Linear(512, 512),
            nn.ReLU(),

            nn.Linear(512, 256),
            nn.ReLU(),

            nn.Linear(256, 10)
        )

    def forward(self, x):
        x = self.flatten(x)
        logits = self.linear_relu_stack(x)
        return logits
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"

print("Using device:", device)

model = NeuralNetwork().to(device)

print(model)
```

```python
# Calculate the total number of trainable parameters
total_params = sum(
    p.numel()
    for p in model.parameters()
    if p.requires_grad
)

print("Total trainable parameters:", total_params)
```

```python
# Autograd verification

x = torch.tensor(2.0, requires_grad=True)
y = torch.tensor(1.0, requires_grad=True)

# Define the function
f = 3 * x**2 + 2 * y**3

# Calculate gradients automatically
f.backward()

# Print the results
print("Function value:", f.item())

print("Expected df/dx: 12")
print("Calculated df/dx:", x.grad.item())

print("Expected df/dy: 6")
print("Calculated df/dy:", y.grad.item())
```

```python
loss_fn = nn.CrossEntropyLoss()

optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.01
)
```

```python
def train(dataloader, model, loss_fn, optimizer):

    model.train()

    for batch, (X, y) in enumerate(dataloader):

        X, y = X.to(device), y.to(device)

        # Forward propagation
        pred = model(X)

        # Calculate loss
        loss = loss_fn(pred, y)

        # Backpropagation
        optimizer.zero_grad()
        loss.backward()

        # Update weights
        optimizer.step()

        if batch % 100 == 0:
            print(f"Batch {batch}: Loss = {loss.item():.4f}")
```

```python
train(train_dataloader, model, loss_fn, optimizer)
```

```python
def test(dataloader, model, loss_fn):

    size = len(dataloader.dataset)
    num_batches = len(dataloader)

    model.eval()

    test_loss = 0
    correct = 0

    with torch.no_grad():

        for X, y in dataloader:

            X, y = X.to(device), y.to(device)

            pred = model(X)

            test_loss += loss_fn(pred, y).item()

            correct += (pred.argmax(1) == y).sum().item()

    test_loss /= num_batches
    correct /= size

    print(f"Test Accuracy: {100 * correct:.2f}%")
    print(f"Average loss: {test_loss:.4f}")
```

```python
test(test_dataloader, model, loss_fn)
```

```python
epochs = 5

for t in range(epochs):

    print(f"\nEpoch {t+1}")
    print("------------------------")

    train(train_dataloader, model, loss_fn, optimizer)

    test(test_dataloader, model, loss_fn)

print("Done!")
```

```python
# Save the trained model weights
torch.save(model.state_dict(), "model_weights.pth")

print("Model weights saved successfully!")
```

```python
# Create a new model
loaded_model = NeuralNetwork().to(device)

# Load the saved weights
loaded_model.load_state_dict(
    torch.load("model_weights.pth",
               map_location=device,
               weights_only=True)
)

# Set the model to evaluation mode
loaded_model.eval()

print("Model loaded successfully!")
```

```python
print("Original model:")
test(test_dataloader, model, loss_fn)

print("\nLoaded model:")
test(test_dataloader, loaded_model, loss_fn)
```

```python
import matplotlib.pyplot as plt

# Select three test images
indices = [0, 1, 2]

fig, axes = plt.subplots(1, 3, figsize=(10, 4))

loaded_model.eval()

with torch.no_grad():

    for ax, idx in zip(axes, indices):

        image, label = test_data[idx]

        # Predict the digit
        prediction = loaded_model(
            image.unsqueeze(0).to(device)
        )

        predicted_label = prediction.argmax(1).item()

        # Display the image
        ax.imshow(image.squeeze(), cmap="gray")

        ax.set_title(
            f"Predicted: {predicted_label}\nActual: {label}"
        )

        ax.axis("off")

plt.tight_layout()
plt.show()
```
